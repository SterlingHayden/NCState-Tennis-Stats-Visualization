## Top 10 singles players

```python
columns_points = ['Singles', 'Singles.1', 'Singles.2', 'Singles.3', 'Singles.4', 'Singles.5', 'Singles.6', 'Singles.7']
columns_players = ['Player 1', 'Player 2', 'Player 3', 'Player 4', 'Player 5', 'Player 6', 'Player 7', 'Player 8']

# Flatten the data across the selected columns
all_values = df_excel[columns_points].values.flatten()
all_players = df_excel[columns_players].values.flatten()

# We will flatten the 'Thru 5/26/2024' column to match the shape of the points and players
all_schools = df_excel['Thru 5/26/2024'].repeat(len(columns_points)).values

# Combine points, players, and school names
points_players_schools = list(zip(all_values, all_players, all_schools))

# Sort by points and get the top 10 values with corresponding players and school names
top_10_points_players_schools = sorted(points_players_schools, key=lambda x: x[0], reverse=True)[:10]

# Convert to DataFrame
df_top_10 = pd.DataFrame(top_10_points_players_schools, columns=['Singles Points', 'Player', 'School'])

# Display the DataFrame
df_top_10
```

```python
df_top_10.to_csv('Data/Data_for_Tableau/top10players_singlespoints.csv', index=False)
```

## National Ranking for Each NCSU Player by Singles, Doubles, and Total Points

```python
columns_doubles_points = ['Doubles', 'Doubles.1', 'Doubles.2', 'Doubles.3', 'Doubles.4', 'Doubles.5', 'Doubles.6', 'Doubles.7']
columns_singles_points = ['Singles', 'Singles.1', 'Singles.2', 'Singles.3', 'Singles.4', 'Singles.5', 'Singles.6', 'Singles.7']
columns_players = ['Player 1', 'Player 2', 'Player 3', 'Player 4', 'Player 5', 'Player 6', 'Player 7', 'Player 8']

# Flatten the data across all universities
doubles_values = df_excel[columns_doubles_points].values.flatten()  # Doubles points across all players
singles_values = df_excel[columns_singles_points].values.flatten()  # Singles points across all players
all_players = df_excel[columns_players].values.flatten()  # Player names across all universities
all_schools = df_excel['Thru 5/26/2024'].repeat(len(columns_doubles_points)).values  # University names, repeated for all players

# Combine doubles points, singles points, players, and school names into a DataFrame
points_players_schools = pd.DataFrame({
    'Player': all_players,
    'School': all_schools,
    'Doubles Points': doubles_values,
    'Singles Points': singles_values
})

# Add a column for Total Points (sum of Doubles and Singles Points)
points_players_schools['Total Points'] = points_players_schools['Doubles Points'] + points_players_schools['Singles Points']

# Rank players by doubles points across all WTA players
points_players_schools['Doubles Rank'] = points_players_schools['Doubles Points'].rank(ascending=False, method='min')

# Rank players by singles points across all WTA players
points_players_schools['Singles Rank'] = points_players_schools['Singles Points'].rank(ascending=False, method='min')

# Rank players by total points across all WTA players
points_players_schools['Total Rank'] = points_players_schools['Total Points'].rank(ascending=False, method='min')

# Filter for NC State players
nc_state_players = points_players_schools[points_players_schools['School'] == 'NC State']

# Sort NC State players by their total points rank, then doubles and singles ranks
nc_state_players_ranked = nc_state_players.sort_values(['Total Rank', 'Doubles Rank', 'Singles Rank'])

# Display the ranked NC State players for doubles, singles, and total points
nc_state_players_ranked
```

```python
nc_state_players_ranked.to_csv('Data/Data_for_Tableau/nc_state_players_ranked.csv', index=False)
```

## Singles and Doubles Points

```python
import pandas as pd

# Assuming 'df_excel' is your original DataFrame

# Define the columns for players, doubles, singles, and total points
columns_doubles_points = ['Doubles', 'Doubles.1', 'Doubles.2', 'Doubles.3', 'Doubles.4', 'Doubles.5', 'Doubles.6', 'Doubles.7']
columns_singles_points = ['Singles', 'Singles.1', 'Singles.2', 'Singles.3', 'Singles.4', 'Singles.5', 'Singles.6', 'Singles.7']
columns_total_points = ['Total Points', 'Total Points.1', 'Total Points.2', 'Total Points.3', 'Total Points.4', 'Total Points.5', 'Total Points.6', 'Total Points.7']
columns_players = ['Player 1', 'Player 2', 'Player 3', 'Player 4', 'Player 5', 'Player 6', 'Player 7', 'Player 8']

# Flatten the data: extracting player names, doubles, singles, and total points
doubles_values = df_excel[columns_doubles_points].values.flatten()  # WTA Doubles Points
singles_values = df_excel[columns_singles_points].values.flatten()  # WTA Singles Points
total_values = df_excel[columns_total_points].values.flatten()  # Total Points
all_players = df_excel[columns_players].values.flatten()  # Player Names
all_schools = df_excel['Thru 5/26/2024'].repeat(len(columns_doubles_points)).values  # Schools

# Create a DataFrame with the required columns
required_columns_df = pd.DataFrame({
    'School': all_schools,         # Schools (one per player)
    'Player': all_players,         # Player names
    'WTA Doubles Points': doubles_values,  # Corresponding doubles points
    'WTA Singles Points': singles_values,  # Corresponding singles points
    'Total Points': total_values   # Total Points by player
})

# Drop rows where 'Player', 'WTA Doubles Points', 'WTA Singles Points', or 'Total Points' are null or 0
required_columns_df = required_columns_df.dropna(subset=['Player', 'WTA Doubles Points', 'WTA Singles Points', 'Total Points'])
required_columns_df = required_columns_df[required_columns_df['Total Points'] > 0]

# Display the result
required_columns_df

# Optionally, save to CSV for Tableau
required_columns_df.to_csv('Data/Data_for_Tableau/singlesDoubles.csv', index=False)
```
